# 描述性统计与可视化

## 目标
对清洗后的数据进行描述性统计与频数可视化。

## 输入
- `data/processed/survey_cleaned.xlsx`

## 输出
- `results/charts/` 下 13 张 PNG 图

## 对应方法
- 均值、中位数、标准差、偏度、峰度
- 频数分布图

```python
# 读取清洗后的数据
import pandas as pd
from pathlib import Path

df = pd.read_excel(Path("../data/processed/survey_cleaned.xlsx"))
print("形状：", df.shape)

# 第 3 题多选题的选项短名（与 01 的编码逻辑一致，用于还原多选计数）
scene_options = ["文娱票务","潮玩周边","户外与娱乐用品","假期旅游","数码电子","服饰美妆","二手交易"]
```

```python
# 描述性统计：月生活费与可自由支配比例
desc = df[["living_num","disposable_num"]].describe().T
desc["方差"] = df[["living_num","disposable_num"]].var().values
desc["标准差"] = df[["living_num","disposable_num"]].std().values
desc["偏度"] = df[["living_num","disposable_num"]].skew().values
desc["峰度"] = df[["living_num","disposable_num"]].kurt().values
print(desc)

living_mean = df["living_num"].mean()
disposable_mean = df["disposable_num"].mean()
print(f"\n生活费均值：{living_mean:.2f} 元")
print(f"可支配比例均值：{disposable_mean:.4f}")
```

```python
# 可视化程序准备
import matplotlib
import matplotlib.pyplot as plt
matplotlib.rcParams['font.family'] = 'sans-serif'
matplotlib.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei']
matplotlib.rcParams['axes.unicode_minus'] = False
```

```python
# 13 张频数分布图
import os
import seaborn as sns
from pathlib import Path

chart_dir = Path("../results/charts")
os.makedirs(chart_dir, exist_ok=True)

def save_show(name):
    plt.tight_layout()
    plt.savefig(chart_dir / f"{name}.png", dpi=150, bbox_inches="tight")
    plt.show()

# 1 年龄
plt.figure(figsize=(7,5))
order = df["age"].value_counts().index.tolist()
sns.countplot(data=df, x="age", order=order, palette="Blues")
plt.title("第1题 年龄分布"); plt.xlabel("年龄"); plt.ylabel("人数")
save_show("01_年龄")

# 2 冲动消费
plt.figure(figsize=(7,5))
order = df["impulse"].value_counts().index.tolist()
sns.countplot(data=df, x="impulse", order=order, palette="Oranges")
plt.title("第2题 冲动消费频率"); plt.xlabel("冲动消费"); plt.ylabel("人数")
plt.xticks(rotation=10)
save_show("02_冲动消费")

# 3 非计划开销场景（多选）
scene_counts = {opt: df[f"scene_{opt}"].sum() for opt in scene_options}
plt.figure(figsize=(9,5))
sns.barplot(x=list(scene_counts.values()), y=list(scene_counts.keys()), palette="viridis")
plt.title(f"第3题 非计划开销场景（多选，n={len(df)}）"); plt.xlabel("选择人数"); plt.ylabel("场景")
save_show("03_非计划开销场景")

# 4 分期意愿
plt.figure(figsize=(7,5))
order = df["demand_4"].value_counts().index.tolist()
sns.countplot(data=df, x="demand_4", order=order, palette="Greens")
plt.title("第4题 分期购买意愿"); plt.xlabel("意愿"); plt.ylabel("人数")
plt.xticks(rotation=10)
save_show("04_分期意愿")

# 5 冷静期
plt.figure(figsize=(7,5))
order = df["cooling"].value_counts().index.tolist()
sns.countplot(data=df, x="cooling", order=order, palette="Set2")
plt.title("第5题 24小时冷静期接受度"); plt.xlabel("态度"); plt.ylabel("人数")
plt.xticks(rotation=10)
save_show("05_冷静期")

# 6 最看重
plt.figure(figsize=(8,5))
order = df["focus"].value_counts().index.tolist()
sns.countplot(data=df, x="focus", order=order, palette="Purples")
plt.title("第6题 分期管理工具最看重项"); plt.xlabel("选项"); plt.ylabel("人数")
plt.xticks(rotation=10)
save_show("06_最看重")

# 7 生活费
plt.figure(figsize=(7,5))
order = df["living"].value_counts().index.tolist()
sns.countplot(data=df, x="living", order=order, palette="Blues")
plt.title("第7题 月生活费分布"); plt.xlabel("生活费区间"); plt.ylabel("人数")
save_show("07_生活费")

# 8 可支配比例
plt.figure(figsize=(7,5))
order = df["disposable"].value_counts().index.tolist()
sns.countplot(data=df, x="disposable", order=order, palette="Greens")
plt.title("第8题 可自由支配比例"); plt.xlabel("可支配比例"); plt.ylabel("人数")
save_show("08_可支配比例")

# 9 分期使用
plt.figure(figsize=(8,5))
order = df["installment_use"].value_counts().index.tolist()
sns.countplot(data=df, x="installment_use", order=order, palette="Oranges")
plt.title("第9题 分期/信用消费使用情况"); plt.xlabel("使用情况"); plt.ylabel("人数")
plt.xticks(rotation=10)
save_show("09_分期使用")

# 10 生活费不够
plt.figure(figsize=(7,5))
order = df["shortage"].value_counts().index.tolist()
sns.countplot(data=df, x="shortage", order=order, palette="Reds")
plt.title("第10题 生活费不够时的应对方式"); plt.xlabel("应对方式"); plt.ylabel("人数")
plt.xticks(rotation=10)
save_show("10_生活费不够")

# 11 额度接受度
plt.figure(figsize=(7,5))
order = df["limit_accept"].value_counts().index.tolist()
sns.countplot(data=df, x="limit_accept", order=order, palette="coolwarm")
plt.title("第11题 额度与用途限制接受度"); plt.xlabel("接受度"); plt.ylabel("人数")
save_show("11_额度接受度")

# 12 分期需求
plt.figure(figsize=(7,5))
order = df["demand"].value_counts().index.tolist()
sns.countplot(data=df, x="demand", order=order, palette="Paired")
plt.title("第12题 分期消费需求"); plt.xlabel("需求"); plt.ylabel("人数")
save_show("12_分期需求")

# 13 官方信任
plt.figure(figsize=(7,5))
order = df["trust"].value_counts().index.tolist()
sns.countplot(data=df, x="trust", order=order, palette="Set3")
plt.title("第13题 官方渠道信任度"); plt.xlabel("信任度"); plt.ylabel("人数")
save_show("13_官方信任")

print("13 张图已保存到：", chart_dir)
```
